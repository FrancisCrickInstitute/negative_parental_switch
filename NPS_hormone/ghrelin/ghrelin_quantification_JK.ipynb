```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ghrelin.csv')
standard = pd.read_csv('standard.csv')
```

```python
sns.set(font_scale=2.7) 
sns.set_style("ticks")
plt.subplots(figsize= (2.7,1.8)) #2.7,4

g = sns.stripplot(x='state', y='Abundance', data = df[df.status == 'third'], order = ['Agg+', 'Agg-', 'Control'], jitter=True, dodge=False, 
              marker='o', size=7, linewidth=1, edgecolor='gray', palette=['red','Tab:blue', 'gray'], alpha=0.5, zorder=1)
g = sns.boxplot(x = 'state', y = 'Abundance', data = df[df.status == 'third'], order = ['Agg+', 'Agg-', 'Control'],
                palette = ['red','Tab:blue', 'gray'], linewidth=1.7, showfliers=False)

plt.ylim(0,1600)
g.yaxis.set_major_locator(plt.MaxNLocator(2))
sns.despine(trim=False, bottom=False)
# g.tick_params(bottom=False)
plt.xlabel(None)
g.set_ylabel('Ghrelin\nabund.',labelpad=10)

g.set_xticklabels(('Agg+','Agg-', 'ctrl'), rotation=45)

for patch in g.artists:
    r, g, b, a = patch.get_facecolor()
    patch.set_facecolor((r, g, b, .5))

plt.savefig('/Volumes/lab-kohlj/home/shared/projects/NegativeParentalSwitch/figures/Figure_4/linked_figure4/ghrelin.pdf',
            dpi=600,bbox_inches='tight', transparent=True)
```
