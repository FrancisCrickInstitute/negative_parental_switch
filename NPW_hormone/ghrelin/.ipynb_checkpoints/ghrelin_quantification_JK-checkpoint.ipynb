```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ghrelin.csv')
standard = pd.read_csv('standard.csv')
```

```python
sns.set(font_scale=2.7) 
sns.set_style("ticks")
plt.subplots(figsize= (3,4))

g = sns.swarmplot(x = 'state', y = 'Abundance', data = df[df.status == 'third'], palette = ['Tab:blue', 'red', 'gray'])
g = sns.boxplot(x = 'state', y = 'Abundance', data = df[df.status == 'third'],  boxprops=dict(alpha=.5), 
                palette = ['Tab:blue', 'red', 'gray'],width = 0.5, showfliers=False)
plt.ylim(0,1500)
g.yaxis.set_major_locator(plt.MaxNLocator(3))
sns.despine(trim=True, bottom=True)
g.tick_params(bottom=False)
plt.xlabel(None)
g.set_xticklabels(('Agg-','Agg+', 'ctrl'), rotation=45)

# plt.savefig('/Users/kohlj/Downloads/ghrelin.pdf',dpi=600,bbox_inches='tight')
plt.savefig('/Volumes/lab-kohlj/home/shared/projects/NegativeParentalSwitch/figures/Figure_4/linked_figure4/ghrelin.pdf',dpi=600,bbox_inches='tight')
```
